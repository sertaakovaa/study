```python
# Ввод слова
word = input("Введите слово: ")

# Проверка длины слова и определение средней буквы
if len(word) % 2 == 0:
    # Если длина чётная, выводим две средние буквы
    middle_chars = word[len(word) // 2 - 1: len(word) // 2 + 1]
else:
    # Если длина нечётная, выводим одну среднюю букву
    middle_chars = word[len(word) // 2]

# Вывод результата
print(f"Средняя буква(ы): {middle_chars}")
```
