```python
from datetime import datetime

# Определение форматов для разных газет
formats = [
    "%A, %B %d, %Y",   # The Moscow Times - Wednesday, October 2, 2002
    "%A, %d.%m.%y",    # The Guardian - Friday, 11.10.13
    "%A, %d %B %Y"     # Daily News - Thursday, 18 August 1977
]

# Функция для конвертации даты
def convert_date(date_str):
    for fmt in formats:
        try:
            # Попытка распарсить строку по каждому формату
            return datetime.strptime(date_str, fmt)
        except ValueError:
            continue
    # Если не удалось преобразовать
    return None

# Итерация по введённым датам
while True:
    date_input = input("Введите дату (или 'exit' для выхода): ")
    if date_input.lower() == 'exit':
        break
    date_obj = convert_date(date_input)
    if date_obj:
        print(f"Конвертированная дата: {date_obj}")
    else:
        print("Ошибка: формат даты не распознан.")
```
