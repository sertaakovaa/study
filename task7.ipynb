```python
import time
from concurrent.futures import ThreadPoolExecutor

# Формула 1
def formula1(x):
    return x**2 - x**2 + x**4 - x**5 + x + x

# Формула 2
def formula2(x):
    return x + x

# Формула 3: сумма результатов формулы 1 и формулы 2
def formula3(result1, result2):
    return result1 + result2

# Функция для выполнения вычислений
def perform_computation(iterations):
    start_time = time.time()

    # Параллельные вычисления формулы 1 и формулы 2
    with ThreadPoolExecutor() as executor:
        # Генерация данных для итераций
        data = range(1, iterations + 1)
        # Параллельный запуск формулы 1
        start_formula1 = time.time()
        results1 = list(executor.map(formula1, data))
        end_formula1 = time.time()

        # Параллельный запуск формулы 2
        start_formula2 = time.time()
        results2 = list(executor.map(formula2, data))
        end_formula2 = time.time()

    # Последовательное выполнение формулы 3
    start_formula3 = time.time()
    results3 = [formula3(r1, r2) for r1, r2 in zip(results1, results2)]
    end_formula3 = time.time()

    end_time = time.time()

    # Вывод времени выполнения
    print(f"Итераций: {iterations}")
    print(f"Время выполнения Формулы 1: {end_formula1 - start_formula1:.4f} секунд")
    print(f"Время выполнения Формулы 2: {end_formula2 - start_formula2:.4f} секунд")
    print(f"Время выполнения Формулы 3: {end_formula3 - start_formula3:.4f} секунд")
    print(f"Общее время: {end_time - start_time:.4f} секунд\n")

# Выполнение задачи
perform_computation(10_000)
perform_computation(100_000)
```
