```python
def move_disk(disk, source, target):
    print(f"Блин {disk}: Стержень {source} -> Стержень {target}.")
    with open("решение.txt", "a") as file:
        file.write(f"Блин {disk}: Стержень {source} -> Стержень {target}.\n")

def hanoi(n, source, target, auxiliary):
    print(f"Выполнение для {n} дисков: {source} -> {target} через {auxiliary}")
    if n == 1:
        move_disk(n, source, target)
    else:
        hanoi(n-1, source, auxiliary, target)  # Переносим n-1 дисков на вспомогательный стержень
        move_disk(n, source, target)  # Переносим сам последний диск
        hanoi(n-1, auxiliary, target, source)  # Переносим n-1 дисков на целевой стержень

# Ввод начальных условий
try:
    disk_count = int(input("Введите количество дисков: "))
    print(f"Количество дисков: {disk_count}")
    rod_count = 3  # Количество стержней для стандартной задачи Ханойской башни
except ValueError:
    print("Ошибка: введено не число! Пожалуйста, введите число.")
    exit()

# Очистим файл перед записью
with open("решение.txt", "w") as file:
    file.write("Решение задачи Ханойская башня:\n")

# Выполнение алгоритма
hanoi(disk_count, 1, 3, 2)
```
