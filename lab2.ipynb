```python
# Ввод товаров и цен
items = input("Введите список товаров через запятую: ").split(", ")
stores = {}

# Запросить цены в магазинах
while True:
    store_name = input("Введите название магазина (или 'стоп' для завершения): ")
    if store_name.lower() == 'стоп':
        break
    store_prices = {}
    for item in items:
        price = float(input(f"Введите цену для товара '{item}' в магазине '{store_name}': "))
        store_prices[item] = price
    stores[store_name] = store_prices

# Рассчитаем общую стоимость покупок в каждом магазине
total_costs = {}
for store, prices in stores.items():
    total_cost = sum(prices.values())
    total_costs[store] = total_cost

# Найдем магазин с минимальной стоимостью
min_cost_store = min(total_costs, key=total_costs.get)

# Вывод результатов
print("\nОбщая стоимость покупок в магазинах:")
for store, total_cost in total_costs.items():
    print(f"{store}: {total_cost:.2f} руб.")

print(f"\nМагазин с наибольшей экономией: {min_cost_store} с общей стоимостью {total_costs[min_cost_store]:.2f} руб.")
```
