```python
def min_breaks(n, m):
    # Если шоколадка уже состоит из одного квадратика, разломов не требуется
    if n == 1 and m == 1:
        return 0
    # Если одна из сторон равна 1, нужно сделать (длина другой стороны - 1) разломов
    elif n == 1 or m == 1:
        return max(n, m) - 1
    # В других случаях делаем разлом по горизонтали или вертикали и применяем рекурсию
    else:
        return (n * m) - 1

# Примеры
print(min_breaks(2, 2))  # 3
print(min_breaks(3, 5))  # 14
print(min_breaks(1, 7))  # 6
```
